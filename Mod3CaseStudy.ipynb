```python
class Vehicle:  # superclass
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type


class Automobile(Vehicle):  # subclass
    def __init__(self, vehicle_type, year, make, model, doors, roof):
        super().__init__(vehicle_type)
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof

    def display_info(self):
        print("\nVehicle type:", self.vehicle_type)
        print("Year:", self.year)
        print("Make:", self.make)
        print("Model:", self.model)
        print("Number of doors:", self.doors)
        print("Type of roof:", self.roof)


def main():
    vehicle_type = "car"
    print("Enter Details for your car")
    year = input("Enter car year: ")
    make = input("Enter make of car: ")
    model = input("Enter model of car: ")
    doors = input("Enter doors of car: ")
    roof = input("Enter roof of car: ")

    user_car = Automobile(vehicle_type, year, make, model, doors, roof)
    user_car.display_info()


if __name__ == "__main__":
    main()
```
